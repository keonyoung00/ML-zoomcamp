````python
df_Asia = df[df['origin'] == 'Asia']
df_Asia.head()
````

````python
df_Asia.loc[:, ['vehicle_weight', 'model_year']]
````

````python
X = df_Asia.loc[:, ['vehicle_weight', 'model_year']][:7].to_numpy()
````

````python
X.T.dot(X)
````

````python
np.linalg.inv(X.T.dot(X))
````

````python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
````

````python
w = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
````
